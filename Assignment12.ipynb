```python
# Diamonds DataFrame Analysis
import pandas as pd

# Create DataFrame (given data)
data = {
    'carat': [0.23, 0.21, 0.23, 0.29, 0.31],
    'cut': ['Ideal', 'Premium', 'Good', 'Premium', 'Good'],
    'color': ['E', 'E', 'E', 'I', 'J'],
    'clarity': ['SI2', 'SI1', 'VS1', 'VS2', 'SI2'],
    'depth': [61.5, 59.8, 56.9, 62.4, 63.3],
    'table': [55.0, 61.0, 65.0, 58.0, 58.0],
    'price': [326, 326, 327, 334, 335],
    'x': [3.95, 3.89, 4.05, 4.20, 4.34],
    'y': [3.98, 3.84, 4.07, 4.23, 4.35],
    'z': [2.43, 2.31, 2.31, 2.63, 2.75]
}

df = pd.DataFrame(data)

# i) Mean price for each cut
print("\nMean Price for each Cut:")
print(df.groupby('cut')['price'].mean())

# ii) Count, Min and Max price for each cut
print("\nCount, Min and Max Price for each Cut:")
print(df.groupby('cut')['price'].agg(['count', 'min', 'max']))

# iii) Average of x, y, z
print("\nAverage values of x, y, z:")
print("x average:", df['x'].mean())
print("y average:", df['y'].mean())
print("z average:", df['z'].mean())
```

```python
# Employee Excel File Analysis
import pandas as pd

# Read Excel file
df = pd.read_excel("employee.xlsx")

# a) Employees in Automotive domain
print("\nEmployees in Automotive Department:")
print(df[df['Department'] == 'Automotive'])

# b) Details using Employee ID
emp_id = int(input("\nEnter Employee ID: "))
print("\nEmployee Details:")
print(df[df['EmployeeID'] == emp_id])

# c) List of Developers
print("\nList of Developers:")
print(df[df['Designation'] == 'Developer'])
```
